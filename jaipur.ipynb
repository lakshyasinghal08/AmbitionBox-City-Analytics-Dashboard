```python
import pandas as pd
import os

# Map every single city dataframe variable to its respective city display name
city_dfs = {
    'Ahmedabad': ahm,
    'Bangalore': ban,
    'Chennai': che,
    'Gurugram': gur,
    'Hyderabad': hyd,
    'Indore': ind,      # Added Indore
    'Jaipur': jai,      # Added Jaipur
    'Mumbai': mum,
    'Noida': noi,
    'Pune': pun
}

combined_list = []

for city_name, df in city_dfs.items():
    # Create a copy to prevent SettingWithCopyWarning
    temp_df = df.copy()
    
    # Drop index column if it exists
    if 'Unnamed: 0' in temp_df.columns:
        temp_df.drop(columns=['Unnamed: 0'], inplace=True)
        
    # Assign the city name column
    temp_df['city'] = city_name
    
    combined_list.append(temp_df)

# Concatenate all into a single master dataframe
master_df = pd.concat(combined_list, ignore_index=True)

# Clean rating column to ensure numeric values
master_df['company_rating'] = pd.to_numeric(master_df['company_rating'], errors='coerce')
master_df['company_rating'] = master_df['company_rating'].fillna(master_df['company_rating'].median())

# Fill remaining missing text values
master_df.fillna('N/A', inplace=True)

# Save to processed folder
os.makedirs('processed_data', exist_ok=True)
output_path = 'processed_data/master_ambitionbox.csv'
master_df.to_csv(output_path, index=False)

print(f"Master file successfully created at {output_path} with {len(master_df)} rows across {master_df['city'].nunique()} cities!")
display(master_df.head())
```
